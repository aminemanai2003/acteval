# Accuracy versus tail calibration
This synthetic example shows why ActEval reports objectives separately.

```python
import numpy as np

import acteval as ae

y = np.r_[np.tile([0.5, 1.0, 1.5, 1.0, 0.5], 19), np.repeat(10.0, 5)]
model_a = np.r_[y[:95] + 0.5, np.repeat(10.0, 5)]
model_b = np.r_[y[:95], np.repeat(9.0, 5)]
```

```python
comparison = ae.compare(
    y,
    {"Model A": model_a, "Model B": model_b},
    task="claim_frequency",
    metrics=["rmse", "poisson_deviance", "tail_ae_95"],
)
comparison.to_dataframe()
```

Model B improves average accuracy while Model A is calibrated in the observed tail.
